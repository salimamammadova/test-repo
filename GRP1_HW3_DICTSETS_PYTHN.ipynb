SETS


```python
'''
1. Find the common elements (intersection) of two sets:

   set1 = {"apple", "book", "cherry", "date"}
   set2 = {"cherry", "date", "book", "grape"}

2. Find the difference between two sets:
   Find the elements that are in the first set but not in the second set.

   set1 = {1, 2, 3, 4, 5}
   set2 = {4, 5, 6, 7, 8}
'''
```

```python
set1 = {"apple", "book", "cherry", "date"}
set2 = {"cherry", "date", "book", "grape"}
print('Intersection = ', set1 & set2)
```

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
print('Difference = ', set1 - set2)
```

Problem-based task

```python
'''
TASK 1:
Write a program to calculate the taxi fare:

Price per kilometer: 0.5 USD is charged for each kilometer driven.
Total price calculation:
Total Price = Distance driven x Price per kilometer.

TASK 2:
Write a program to calculate daily water intake:

A person should drink 30 ml of water per kilogram of body weight per day.
The user enters their weight in kilograms.
Daily water amount calculation:
Water amount = Weight x 30 ml.
'''
```

```python
distance = float(input('Enter distance in km: '))
rate_per_km = 0.5
total_price = distance * rate_per_km
print('Total price:', total_price)
```

```python
weight = float(input('Enter your weight in kg: '))
ml_per_kg = 30
total_water_ml = weight * ml_per_kg
water_liters = total_water_ml / 1000
print('Daily water requirement:', water_liters, 'L')
```
