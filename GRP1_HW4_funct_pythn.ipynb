```python
"""
Task 1: Ask the user for speed (v) and distance (s). Then calculate time (t).
Formula: t = s / v
v is in meters/second
s is in meters
Result: Output the time in hours.

Task 2: Calculate 50% Area Increase:
Task: A square has a side length of 6 cm. Increase the area of the square by 50%.
Write code to find the new area.
"""
```

```python
#Task1:Calculate time in hours from speed (m/s) and distance (m)
v = float(input("Enter speed (m/s): "))
s = float(input("Enter distance (meters): "))

# Formula:t=s/v
t_seconds = s / v

#1 hour = 3600 seconds
t_hours = t_seconds / 3600

print(f"Time in seconds: {t_seconds} s")
print(f"Time in hours: {t_hours} hours")
```

```python
# Task 2: Calculate the new area of a square increased by 50%
side_length = 6

#Calculate the original area (S = side^2)
original_area = side_length ** 2

#Increase the area by 50% (multiply by 1.5)
new_area = original_area * 1.5

print(f"Original Area: {original_area} cm²")
print(f"New Area (50% increased): {new_area} cm²")
```
